# Data Cleaning Process Overview

#### This series of code snippets is part of the data cleaning process for LinkedIn data stored in CSV files. The tasks involve renaming files, removing duplicates, handling special characters, encoding issues, and cleaning column headers. Each part of the process is carefully designed to ensure the data is properly structured and free from inconsistencies before further analysis.

## 1. File Renaming Process

### Explanation:

   #### Renaming CSV Files: The script iterates through each CSV file in the specified directory.

   #### Cleaning Filenames: Underscores (_) and dashes (-) are replaced with spaces. Additionally, multiple spaces are compressed into a single space.

   #### Duplicate Name Check: It checks if the first half of the name is identical to the second half (ignoring case), and if so, truncates the filename to just the first half.

   #### File Renaming: The new cleaned filenames are applied and files are renamed accordingly.

```python
import os

for filename in os.listdir("LinkedIn Data Public"):
    file_path = os.path.join("LinkedIn Data Public", filename)

    if os.path.isfile(file_path):
        name, ext = os.path.splitext(filename)

        # Replace underscore and dash with space
        cleaned = name.replace("_", " ").replace("-", " ")

        # Remove extra spaces
        cleaned = " ".join(cleaned.split())

        # Check for exact duplicate even if capital letters alag hain
        words = cleaned.split()
        half = len(words) // 2
        if len(words) >= 2 and [w.lower() for w in words[:half]] == [w.lower() for w in words[half:]]:
            cleaned = " ".join(words[:half])

        # Final filename
        new_filename = cleaned + ".csv"
        new_path = os.path.join("LinkedIn Data Public", new_filename)

        os.rename(file_path, new_path)
        print(f"Renamed: {filename} → {new_filename}")
```

### Explanation:

#### File Reading: The code attempts to read CSV files with different encodings (UTF-8  and ISO-8859-1), skipping any problematic lines.

####    Column Cleaning: It strips whitespace from column headers to ensure consistency.

####    Data Merging: If the necessary columns (First Name, Last Name, Company) are present, it creates a new column called Name by merging the first and last name.

####    Data Output: A new DataFrame with only the Name and Company columns is saved back to the file.

```python
import os
import pandas as pd

folder_path = 'LinkedIn Data Public'

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)

        # Try reading with utf-8, ISO-8859-1, and handle parser issues
        try:
            try:
                df = pd.read_csv(file_path, encoding='utf-8', on_bad_lines='skip')
            except UnicodeDecodeError:
                df = pd.read_csv(file_path, encoding='ISO-8859-1', on_bad_lines='skip')
        except Exception as e:
            print(f" Skipped (cannot read): {filename} → {e}")
            continue

        # Strip any whitespace from column headers
        df.columns = df.columns.str.strip()

        # Check if required columns are there
        if 'First Name' in df.columns and 'Last Name' in df.columns and 'Company' in df.columns:
            # Create merged name
            df['Name'] = df['First Name'].astype(str) + ' ' + df['Last Name'].astype(str)

            # Create new DataFrame with only required columns
            new_df = df[['Name', 'Company']]

            # Save back to same file
            new_df.to_csv(file_path, index=False)
            print(f" Updated: {filename}")
        else:
            print(f" {filename}")
```

## Handling Encoding and Parsing Issues

### Explanation:

####    Error Handling: This script handles files that might be unreadable due to encoding or parsing issues. If an error is encountered, the file is skipped with a message.

####     File Details: It prints details about the file, including the number of columns and the column names, providing insights into the structure of the data.



```python
import os
import pandas as pd

folder_path = 'LinkedIn Data Public'

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)

        try:
            # Attempt to read the file
            df = pd.read_csv(file_path, encoding='utf-8', on_bad_lines='skip')
        except (UnicodeDecodeError, pd.errors.ParserError):
            # If error occurs (encoding or parsing), skip this file
            print(f"⚠️ Skipping file: {filename} due to encoding or parsing error.")
            continue

        # Print file name and column details
        print(f"File: {filename}")
        print(f"Number of Columns: {len(df.columns)}")
        print(f"Column Names: {df.columns.tolist()}")
        print("-" * 50)
```

## 4. Cleaning Non-ASCII Characters and Special Characters

### Explanation:

####    Text Cleaning: The clean_text function removes non-ASCII characters, special characters, and extra spaces from the text.

####     Data Processing: For each file, it checks if the required columns (name, company name) exist. If they do, it applies the cleaning function to both columns and stores the cleaned data in a new folder.

####    Folder Creation: The script ensures that the output folder exists before saving the cleaned files.

```python
import os
import pandas as pd
import re

folder_path = 'LinkedIn Data Public'
output_folder = 'Cleaned_Files'
os.makedirs(output_folder, exist_ok=True)

def clean_text(text):
    if pd.isna(text):
        return ''
    text = text.encode('ascii', 'ignore').decode('ascii')  # Remove emojis & non-ASCII
    text = re.sub(r'[^\w\s]', '', text)  # Remove all special characters except space
    text = re.sub(r'\s+', ' ', text)  # Remove multiple spaces
    return text.strip()

cleaned_count = 0

for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)

        try:
            df = pd.read_csv(file_path, encoding='utf-8', on_bad_lines='skip')
        except (UnicodeDecodeError, pd.errors.ParserError):
            continue  # skip unreadable files

        available_cols = [col.lower().strip() for col in df.columns]
        if 'name' in available_cols and 'company name' in available_cols:
            name_col = df.columns[available_cols.index('name')]
            company_col = df.columns[available_cols.index('company name')]

            df_cleaned = df[[name_col, company_col]].copy()
            df_cleaned[name_col] = df_cleaned[name_col].apply(clean_text)
            df_cleaned[company_col] = df_cleaned[company_col].apply(clean_text)

            cleaned_path = os.path.join(output_folder, filename)
            df_cleaned.to_csv(cleaned_path, index=False)
            cleaned_count += 1
            print(f" Cleaned: {filename}")
        # else:
        #     print(f" Skipped (columns missing): {filename}")
```

```python
import os
import pandas as pd
import re

# Folder path
folder_path = 'LinkedIn Data Public'

# Function to clean data
def clean_string_format(df):
    # Loop through each column and clean each value
    for col in df.columns:
        df[col] = df[col].apply(lambda x: re.sub(r'[^\x00-\x7F]+', '', str(x)))  # Remove non-ASCII characters
        df[col] = df[col].apply(lambda x: re.sub(r'\"', '', str(x)))  # Remove quotation marks
        df[col] = df[col].apply(lambda x: re.sub(r'\s+', ' ', str(x)))  # Remove extra spaces
    return df

# Read all files and clean them
for filename in os.listdir(folder_path):
    if filename.endswith('.csv'):
        file_path = os.path.join(folder_path, filename)
        try:
            # Read the CSV file
            df = pd.read_csv(file_path, encoding='utf-8', on_bad_lines='skip')
            # Clean the string formatting in the DataFrame
            df = clean_string_format(df)
            # Save the cleaned file back
            df.to_csv(file_path, index=False)
            print(f"{filename} cleaned successfully.")
        except Exception as e:
            print(f" Error with {filename}: {str(e)}")
```

## 5. Handling Encoding Detection

### Explanation:

####    Encoding Detection: This code uses the chardet library to automatically detect the encoding of each CSV file, ensuring it reads files without issues regardless of encoding.

####   Data Cleaning: It normalizes column names (removes extra spaces and converts to lowercase), removes fully empty rows, and sorts the data based on the name column.

####   Final Output: After sorting, the cleaned data is saved back to the same file.

```python
import os
import pandas as pd

folder_path = "LinkedIn Data Public"
encodings_to_try = ['utf-8', 'ISO-8859-1', 'utf-16', 'latin1']

def read_csv_safely(file_path):
    for enc in encodings_to_try:
        try:
            df = pd.read_csv(file_path, encoding=enc, on_bad_lines='skip')
            print(f" {os.path.basename(file_path)} read successfully using encoding: {enc}")
            return df
        except Exception as e:
            continue  # silently try next encoding
    print(f"Skipped: {os.path.basename(file_path)} - No compatible encoding found.")
    return None

# Go through all files in the folder and attempt to read them
for filename in os.listdir(folder_path):
    if filename.endswith(".csv"):
        file_path = os.path.join(folder_path, filename)
        df = read_csv_safely(file_path)
```

## 6. Sorting By Name 

### Sorting by 'name':

#### The DataFrame is sorted alphabetically by the 'name' column in a case-insensitive manner, with missing values pushed to the bottom.

```python
import os
import pandas as pd
import chardet

folder_path = "LinkedIn Data Public"

for filename in os.listdir(folder_path):
    if filename.endswith(".csv"):
        file_path = os.path.join(folder_path, filename)
        try:
            # Detect encoding
            with open(file_path, 'rb') as f:
                result = chardet.detect(f.read())
            encoding = result['encoding']

            # Read the file
            df = pd.read_csv(file_path, encoding=encoding, on_bad_lines='skip')

            # Normalize column names
            df.columns = df.columns.str.strip().str.lower()

            # Remove fully empty rows
            df = df.dropna(how='all')

            # Strip whitespace from 'name' column if it exists
            if 'name' in df.columns:
                df['name'] = df['name'].astype(str).str.strip()
                
                # Sort by name (case insensitive), push blanks/NaNs to bottom
                df_sorted = df.sort_values(by='name', key=lambda x: x.str.lower(), na_position='last', ignore_index=True)
                df_sorted.to_csv(file_path, index=False)
                print(f" Sorted by 'name': {filename}")
            else:
                print(f"⚠️ 'name' column missing in file: {filename}")

        except Exception as e:
            print(f" Skipping file: {filename} due to error: {e}")

print("\nAll files processed.")
```
